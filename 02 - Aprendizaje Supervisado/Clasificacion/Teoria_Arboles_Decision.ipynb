# 🌳 Teoría de Árboles de Decisión

---

## 📚 1. Introducción

Los **Árboles de Decisión** son algoritmos de **aprendizaje supervisado** que modelan decisiones mediante una estructura de árbol jerárquica. Son útiles tanto para **clasificación** como para **regresión**.

### 🏗️ Estructura de un Árbol de Decisión

```
                    [Nodo Raíz]
                   (feature X1)
                    /        \\
                   /          \\
              X1 < 5         X1 >= 5
              /                  \\
        [Nodo Interno]      [Nodo Interno]
         (feature X2)        (feature X3)
          /      \\             /      \\
        /         \\           /         \\
    [Hoja:A]  [Hoja:B]  [Hoja:C]  [Hoja:D]
```

### 🔑 Componentes Clave:

| Componente | Descripción | Función |
|------------|-------------|----------|
| 🌱 **Nodo Raíz** | Primer nodo del árbol | Contiene todos los datos iniciales |
| 🔵 **Nodos Internos** | Nodos de decisión | Evalúan condiciones sobre features |
| ➡️ **Ramas** | Conexiones entre nodos | Representan el resultado de una decisión |
| 🍃 **Nodos Hoja** | Nodos finales | Contienen la predicción final (clase o valor) |

---

### 🧮 Definición Formal

Un árbol de decisión es una función $h: \\mathbb{X} \\rightarrow \\mathbb{Y}$ que particiona recursivamente el espacio de entrada $\\mathbb{X}$ mediante reglas de decisión:

$$h(\\mathbf{x}) = \\begin{cases}
y_1 & \\text{if } \\mathbf{x} \\in R_1 \\\\
y_2 & \\text{if } \\mathbf{x} \\in R_2 \\\\
\\vdots \\\\
y_k & \\text{if } \\mathbf{x} \\in R_k
\\end{cases}$$

**Donde:**
* $R_1, R_2, ..., R_k$ son **regiones** del espacio de características
* $y_i$ es la **predicción** para la región $R_i$
* El árbol divide el espacio en regiones **disjuntas** (sin solapamiento)

## ⚙️ 2. Algoritmo de Construcción del Árbol

---

### 💻 Algoritmo ID3 (Iterative Dichotomiser 3)

Desarrollado por **Ross Quinlan** en 1986, ID3 construye el árbol de manera **top-down** (de arriba hacia abajo) utilizando **Ganancia de Información**.

#### 📝 Pseudocódigo:

```python
función ID3(ejemplos, atributos, clase_objetivo):
    """
    Construye un árbol de decisión recursivamente
    
    Parámetros:
        ejemplos: Dataset de entrenamiento
        atributos: Features disponibles para dividir
        clase_objetivo: Variable a predecir
    """
    
    # CASO BASE 1: Todos los ejemplos son de la misma clase
    if todos_ejemplos_misma_clase(ejemplos):
        return crear_hoja(clase_mayoritaria)
    
    # CASO BASE 2: No quedan atributos para dividir
    if len(atributos) == 0:
        return crear_hoja(clase_mayoritaria(ejemplos))
    
    # PASO RECURSIVO: Seleccionar mejor atributo
    mejor_atributo = seleccionar_mejor_atributo(ejemplos, atributos)
    
    # Crear nodo para mejor_atributo
    arbol = crear_nodo(mejor_atributo)
    
    # Para cada valor posible del mejor atributo
    for valor in valores_posibles(mejor_atributo):
        # Dividir ejemplos según el valor
        ejemplos_subconjunto = filtrar(ejemplos, mejor_atributo == valor)
        
        # Recursión: Construir subárbol
        if ejemplos_subconjunto está vacío:
            agregar_hoja(arbol, clase_mayoritaria(ejemplos))
        else:
            subarbol = ID3(ejemplos_subconjunto, 
                          atributos - {mejor_atributo}, 
                          clase_objetivo)
            agregar_rama(arbol, valor, subarbol)
    
    return arbol
```

#### 🔄 Flujo del Algoritmo:

```
1. Calcular entropía del nodo actual
   ↓
2. Para cada atributo:
   - Calcular Ganancia de Información
   ↓
3. Seleccionar atributo con MAYOR ganancia
   ↓
4. Dividir dataset según valores del atributo
   ↓
5. Aplicar ID3 recursivamente a cada subconjunto
   ↓
6. CASO BASE: Si todos ejemplos son de la misma clase → HOJA
```

---

### 🎯 Función: Seleccionar Mejor Atributo

```python
función seleccionar_mejor_atributo(ejemplos, atributos):
    """
    Selecciona el atributo que maximiza la Ganancia de Información
    """
    mejor_ganancia = -∞
    mejor_atributo = None
    
    # Calcular entropía del conjunto actual
    entropia_padre = calcular_entropia(ejemplos)
    
    for atributo en atributos:
        # Calcular ganancia de información
        ganancia = calcular_ganancia_informacion(ejemplos, atributo, entropia_padre)
        
        if ganancia > mejor_ganancia:
            mejor_ganancia = ganancia
            mejor_atributo = atributo
    
    return mejor_atributo
```

## 📊 3. Entropía y Ganancia de Información

---

### 3.1 🌪️ Entropía

La **entropía** mide la **impureza** o **desorden** de un conjunto de datos. Concepto tomado de la teoría de la información de Claude Shannon.

#### 🧮 Fórmula:

$$H(S) = -\\sum_{i=1}^{c} p_i \\log_2(p_i)$$

**Donde:**
* $S$ = Conjunto de datos
* $c$ = Número de clases
* $p_i$ = Proporción de ejemplos de la clase $i$

#### 📈 Propiedades:

| Condición | Entropía | Interpretación |
|------------|----------|----------------|
| Todos los ejemplos de la **misma clase** | $H(S) = 0$ | ✅ **Máxima pureza** (sin desorden) |
| Ejemplos **distribuidos uniformemente** | $H(S) = \\log_2(c)$ | ❌ **Máxima impureza** (máximo desorden) |
| Distribución **intermedia** | $0 < H(S) < \\log_2(c)$ | 🟡 **Impureza parcial** |

#### 💡 Ejemplo Práctico:

**Dataset:**
* 10 ejemplos Positivos (+)
* 5 ejemplos Negativos (-)
* Total: 15 ejemplos

**Cálculo:**

$$p_+ = \\frac{10}{15} = 0.67, \\quad p_- = \\frac{5}{15} = 0.33$$

$$H(S) = -0.67 \\log_2(0.67) - 0.33 \\log_2(0.33) = 0.918$$

📈 **Interpretación**: Entropía alta (cerca de 1) indica **desorden** → necesitamos más información para clasificar.

---

### 3.2 🌟 Ganancia de Información (Information Gain)

Mide cuánto **reduce la entropía** un atributo al dividir el dataset.

#### 🧮 Fórmula:

$$IG(S, A) = H(S) - \\sum_{v \\in Values(A)} \\frac{|S_v|}{|S|} H(S_v)$$

**Donde:**
* $A$ = Atributo candidato
* $Values(A)$ = Valores posibles del atributo $A$
* $S_v$ = Subconjunto de $S$ donde $A = v$
* $|S_v|$ = Número de ejemplos en $S_v$

#### 📊 Interpretación:

* **Alta ganancia** (↑) = El atributo **divide bien** los datos → ✅ Buen split
* **Baja ganancia** (↓) = El atributo **NO ayuda** a separar clases → ❌ Mal split

#### 💡 Ejemplo Práctico:

**Dataset de Juego de Tenis:**

| Outlook | Temp | Humidity | Wind | Play? |
|---------|------|----------|------|-------|
| Sunny | Hot | High | Weak | No |
| Sunny | Hot | High | Strong | No |
| Overcast | Hot | High | Weak | Yes |
| Rain | Mild | High | Weak | Yes |
| Rain | Cool | Normal | Weak | Yes |

**Calcular Ganancia para Outlook:**

1. Entropía total: $H(S) = 0.971$
2. Dividir por Outlook:
   * Sunny: 2 No, 0 Yes → $H(Sunny) = 0$
   * Overcast: 0 No, 1 Yes → $H(Overcast) = 0$
   * Rain: 0 No, 2 Yes → $H(Rain) = 0$
3. Ganancia: $IG(S, Outlook) = 0.971 - 0 = 0.971$ ✅ **¡Excelente split!**

---

### 🎯 Selección del Mejor Atributo

**Criterio**: Elegir el atributo con **MAYOR Ganancia de Información**.

```python
mejor_atributo = max(atributos, key=lambda A: IG(S, A))
```

## 📏 4. Índice Gini

---

### 4.1 🔵 Impureza de Gini

El **Índice de Gini** es otra medida de impureza, más simple computacionalmente que la entropía. Utilizado en el algoritmo **CART** (Classification and Regression Trees).

#### 🧮 Fórmula:

$$\\text{Gini}(S) = 1 - \\sum_{i=1}^{c} p_i^2$$

**Donde:**
* $p_i$ = Proporción de ejemplos de la clase $i$
* Se calcula como la probabilidad de **clasificar incorrectamente** un ejemplo al azar

#### 📈 Propiedades:

| Condición | Gini | Interpretación |
|------------|------|----------------|
| Todos los ejemplos de la **misma clase** | $\\text{Gini}(S) = 0$ | ✅ **Máxima pureza** |
| Ejemplos **distribuidos uniformemente** | $\\text{Gini}(S) = 0.5$ | ❌ **Máxima impureza** |
| Distribución **intermedia** | $0 < \\text{Gini}(S) < 0.5$ | 🟡 **Impureza parcial** |

#### 💡 Ejemplo Práctico:

**Dataset:**
* 10 ejemplos Positivos (+)
* 5 ejemplos Negativos (-)
* Total: 15 ejemplos

**Cálculo:**

$$p_+ = \\frac{10}{15} = 0.67, \\quad p_- = \\frac{5}{15} = 0.33$$

$$\\text{Gini}(S) = 1 - (0.67^2 + 0.33^2) = 1 - (0.449 + 0.109) = 0.442$$

---

### 4.2 ⚖️ Entropía vs Gini

| Criterio | Entropía | Gini |
|----------|----------|------|
| **Fórmula** | $-\\sum p_i \\log_2(p_i)$ | $1 - \\sum p_i^2$ |
| **Rango** | $[0, \\log_2(c)]$ | $[0, 0.5]$ (binario) |
| **Complejidad computacional** | 🐌 Alta (logaritmo) | ⚡ Baja (cuadrática) |
| **Algoritmo** | ID3, C4.5 | CART |
| **Sensibilidad** | Más sensible a cambios | Menos sensible |
| **Resultado** | 🔄 Generalmente similares | 🔄 Generalmente similares |

#### 🎯 ¿Cuál Usar?

* **Entropía**: Interpretación teórica más sólida (teoría de la información)
* **Gini**: Más rápido de computar, suficientemente preciso
* **En la práctica**: Ambos producen árboles similares

**PySpark ML usa Gini por defecto** (⚡ más eficiente).

---

### 4.3 📊 Reducción de Gini (Gini Gain)

Análogo a la Ganancia de Información, pero usando Gini:

$$\\Delta\\text{Gini}(S, A) = \\text{Gini}(S) - \\sum_{v \\in Values(A)} \\frac{|S_v|}{|S|} \\text{Gini}(S_v)$$

**Criterio de selección**: Elegir el atributo que **maximiza** la reducción de Gini.

```python
mejor_atributo = max(atributos, key=lambda A: ΔGini(S, A))
```

## ✂️ 5. Poda (Pruning) y Control de Overfitting

---

### ⚠️ Problema del Overfitting

Los árboles de decisión tienden a **sobreajustarse** (overfitting):

* 🌳 **Árbol muy profundo** → Memoriza los datos de entrenamiento (incluyendo **ruido**)
* 📉 **Alta precisión en train** pero **baja precisión en test**
* 🚫 **No generaliza** a datos nuevos

#### 📊 Ejemplo Visual:

```
Train Accuracy: 100%   ✅ ¡Perfecto!
Test Accuracy:   65%   ❌ Overfitting
```

---

### 5.1 🚫 Poda Pre-procesamiento (Pre-pruning / Early Stopping)

**Detener el crecimiento del árbol ANTES de que se complete**, estableciendo límites:

| Parámetro PySpark | Descripción | Valor Recomendado | Efecto |
|-------------------|-------------|-------------------|---------|
| `maxDepth` | Profundidad máxima del árbol | 5-15 | Evita árboles muy profundos |
| `minInstancesPerNode` | Mínimo de muestras por nodo | 5-20 | Evita hojas con muy pocos ejemplos |
| `maxBins` | Número máximo de bins para discretizar features continuos | 32 (default) | Controla complejidad |
| `minInfoGain` | Ganancia mínima requerida para dividir | 0.0 (default) | Solo divide si mejora sustancialmente |

#### 💻 Ejemplo en PySpark:

```python
from pyspark.ml.classification import DecisionTreeClassifier

dt = DecisionTreeClassifier(
    featuresCol='features',
    labelCol='label',
    maxDepth=10,                # Profundidad máxima: 10 niveles
    minInstancesPerNode=5,      # Mínimo 5 muestras por nodo
    minInfoGain=0.01           # Ganancia mínima: 0.01
)
```

---

### 5.2 ✂️ Poda Post-procesamiento (Post-pruning)

**Construir el árbol completo y LUEGO podar** ramas que no mejoran el rendimiento:

#### 🔄 Proceso:

```
1. Entrenar árbol completo (sin restricciones)
   ↓
2. Para cada nodo interno (de abajo hacia arriba):
   - Evaluar accuracy EN VALIDATION SET con y sin el nodo
   ↓
3. Si eliminar el nodo MEJORA o MANTIENE accuracy:
   - ✂️ Podar (convertir en hoja)
   ↓
4. Repetir hasta que no haya más mejoras
```

#### ⚖️ Ventajas vs Pre-pruning:

| Característica | Pre-pruning | Post-pruning |
|-----------------|-------------|---------------|
| **Cuándo se aplica** | Durante construcción | Después de construcción |
| **Complejidad** | ⚡ Menor | 🐌 Mayor |
| **Riesgo** | Puede sub-ajustar | Menor riesgo |
| **PySpark** | ✅ Soportado | ❌ No soportado directamente |
| **Sklearn** | ✅ Soportado | ✅ Soportado (`cost_complexity_pruning`) |

💡 **PySpark**: Solo soporta **pre-pruning**. Para post-pruning, usar sklearn o implementar manualmente.

---

### 5.3 📊 Estrategias de Validación

#### 🔄 Cross-Validation

```python
from pyspark.ml.tuning import CrossValidator, ParamGridBuilder
from pyspark.ml.evaluation import BinaryClassificationEvaluator

# Crear grid de hiperparámetros
paramGrid = ParamGridBuilder() \\
    .addGrid(dt.maxDepth, [5, 10, 15, 20]) \\
    .addGrid(dt.minInstancesPerNode, [1, 5, 10]) \\
    .build()

# Configurar cross-validation
cv = CrossValidator(
    estimator=dt,
    estimatorParamMaps=paramGrid,
    evaluator=BinaryClassificationEvaluator(),
    numFolds=5,          # 5-fold CV
    seed=42
)

# Entrenar con CV
cvModel = cv.fit(train_data)

# Mejor modelo
best_model = cvModel.bestModel
print(f"Mejor maxDepth: {best_model.getMaxDepth()}")
print(f"Mejor minInstancesPerNode: {best_model.getMinInstancesPerNode()}")
```

---

### 🎯 Conclusión sobre Poda

✅ **Usar siempre pre-pruning** en PySpark mediante:
* `maxDepth` (más importante)
* `minInstancesPerNode`
* Validar con **Cross-Validation** para encontrar mejores valores

❌ **Evitar**: Árboles sin restricciones (`maxDepth` muy alto) en datos reales

## ⚖️ 6. Ventajas y Desventajas

---

### ✅ Ventajas de los Árboles de Decisión

| Ventaja | Descripción | Impacto |
|---------|-------------|--------|
| 📄 **Interpretabilidad** | Fáciles de visualizar y explicar a no-expertos | 🎯 Excelente para casos de negocio |
| 🚫 **No paramétricos** | Sin supuestos sobre distribución de datos | 🚀 Funcionan con cualquier distribución |
| 🔢 **Múltiples tipos de datos** | Aceptan features numéricos y categóricos | 🔧 No requiere one-hot encoding |
| 🏆 **Feature Importance** | Identifican automáticamente las variables más importantes | 💡 Ayuda en selección de features |
| 📊 **No-linealidad** | Capturan relaciones no-lineales sin transformación | 🌀 Modelos complejos sin ingeniería manual |
| ⚡ **Predicción rápida** | Tiempo de predicción es $O(\\log n)$ | ⏱️ Muy rápido en producción |
| 🛡️ **Robusto a outliers** | No afectado por valores extremos | 💪 Funciona bien con datos ruidosos |
| 🚫 **No requiere normalización** | Invariante a escalas de features | 🎉 Menos preprocesamiento |

---

### ❌ Desventajas de los Árboles de Decisión

| Desventaja | Descripción | Solución |
|-----------|-------------|----------|
| 🔴 **Overfitting** | Tienden a memorizar datos de entrenamiento | ✂️ Poda, limitar `maxDepth` |
| 📉 **Inestabilidad** | Pequeños cambios en datos → árbol completamente diferente | 🌲 Usar **Random Forest** |
| ⚠️ **Sesgos con clases desbalanceadas** | Favorece la clase mayoritaria | ⚖️ Balancear clases o usar pesos |
| 🔵 **Splits sub-óptimos** | Greedy algorithm: no garantiza el mejor árbol global | 🧠 Usar **ensemble methods** |
| 📊 **Relaciones lineales simples** | Puede ser ineficiente para relaciones lineales | 📈 Mejor usar regresión lineal |
| 💻 **Extrapolación pobre** | No predice fuera del rango de entrenamiento | ⚠️ No usar para extrapolación |
| 🧩 **XOR y problemas complejos** | Dificultad con patrones XOR (requiere múltiples splits) | 🤝 Combinar múltiples árboles |

---

### 🎯 Cuándo Usar Árboles de Decisión

#### ✅ Casos de Uso Ideales:

* 🏦 **Clasificación de crédito**: Interpretar por qué se rechazó un préstamo
* 🏥 **Diagnóstico médico**: Explicar decisiones a pacientes
* 📊 **Segmentación de clientes**: Identificar reglas de negocio
* 🛡️ **Detección de fraude**: Interpretar por qué una transacción es sospechosa
* 🌱 **Prototipado rápido**: Obtener baseline rápido

#### ❌ Cuándo NO Usar:

* 📈 **Relaciones lineales simples** → Mejor: Regresión Lineal
* 📊 **Extrapolación** → Mejor: Modelos paramétricos
* 🏆 **Máxima precisión** → Mejor: **Random Forest, Gradient Boosting**
* 🕹️ **Imágenes, texto** → Mejor: Deep Learning

---

### 🔄 Evolución: De Árboles Individuales a Ensembles

```
🌳 Decision Tree
    ↓
    Problema: Overfitting, Inestabilidad
    ↓
🌲 Random Forest (Bootstrap + Bagging)
    ↓
    Problema: Convergencia lenta
    ↓
🚀 Gradient Boosted Trees (XGBoost, LightGBM)
    ↓
    🏆 Estado del arte en tabular data
```

💡 **Recomendación**: Usar Árboles individuales para **interpretabilidad**, Random Forest o GBT para **performance**.

# Aprendizaje por Refuerzo (Reinforcement Learning)

## 🎯 Definición

**Aprendizaje por Refuerzo** (Reinforcement Learning, RL) es un paradigma de Machine Learning donde un **agente** aprende a tomar **decisiones** interactuando con un **entorno**, recibiendo **recompensas** (o penalizaciones) por sus acciones, con el objetivo de **maximizar la recompensa acumulada** a largo plazo.

### Diferencia con otros paradigmas

| Paradigma | Datos | Objetivo | Feedback |
|-----------|-------|----------|----------|
| **Supervisado** | Etiquetados | Predecir etiqueta correcta | Inmediato (correcto/incorrecto) |
| **No Supervisado** | Sin etiquetas | Descubrir estructura | Sin feedback explícito |
| **Por Refuerzo** | Interacción | Maximizar recompensa acumulada | Diferido, parcial, escaso |

### Características clave

* 🤖 **Agente autónomo**: Toma decisiones secuenciales
* 🌍 **Entorno dinámico**: Responde a las acciones del agente
* 💰 **Recompensas diferidas**: El resultado de una acción puede verse mucho después
* ⚖️ **Trade-off exploración/explotación**: Balance entre explorar opciones nuevas y explotar conocimiento actual
* 📈 **Aprendizaje por prueba y error**: El agente mejora mediante experiencia

## 📘 Conceptos Fundamentales

### Componentes de un sistema de RL

```
     ┌─────────────┐
     │   Entorno   │
     │  (Mundo)    │
     └──────┬──────┘
            │ Estado (s_t)
            │ Recompensa (r_t)
            ↓
     ┌─────────────┐
     │   Agente    │
     │  (Decisor)  │
     └──────┬──────┘
            │ Acción (a_t)
            ↓
     [Ciclo se repite]
```

1. **Agente**: Entidad que toma decisiones (ej: robot, jugador de ajedrez, sistema de trading)
2. **Entorno**: Mundo con el que interactúa el agente (ej: tablero, mercado financiero)
3. **Estado** ($s_t$): Representación de la situación actual del entorno en el tiempo $t$
4. **Acción** ($a_t$): Decisión que toma el agente en el estado $s_t$
5. **Recompensa** ($r_t$): Señal numérica que el entorno da al agente tras una acción
6. **Política** ($\pi$): Estrategia que mapea estados a acciones: $\pi(a|s)$
7. **Función de valor**: Estima la "bondad" de un estado o par (estado, acción)

---

## 🔄 Proceso Markoviano de Decisión (MDP)

RL se modela típicamente como un **MDP (Markov Decision Process)**:

$$\text{MDP} = \langle \mathcal{S}, \mathcal{A}, P, R, \gamma \rangle$$

Donde:
* $\mathcal{S}$: Conjunto de estados
* $\mathcal{A}$: Conjunto de acciones
* $P(s'|s, a)$: Probabilidad de transición al estado $s'$ dado estado $s$ y acción $a$
* $R(s, a)$: Recompensa esperada al tomar acción $a$ en estado $s$
* $\gamma \in [0, 1]$: Factor de descuento (importancia de recompensas futuras)

### Propiedad de Markov

El futuro depende **solo del estado presente**, no del pasado:

$$P(s_{t+1} | s_t, a_t, s_{t-1}, a_{t-1}, ..., s_0, a_0) = P(s_{t+1} | s_t, a_t)$$

Esto simplifica el problema: el estado $s_t$ contiene toda la información relevante.

---

## 💡 Objetivo del Agente

El objetivo es encontrar una **política óptima** $\pi^*$ que maximice el **retorno esperado** $G_t$ (recompensa acumulada descontada):

$$G_t = r_{t+1} + \gamma r_{t+2} + \gamma^2 r_{t+3} + ... = \sum_{k=0}^{\infty} \gamma^k r_{t+k+1}$$

Donde:
* $r_{t+1}$: Recompensa inmediata en $t+1$
* $\gamma$: Factor de descuento ($0 \leq \gamma < 1$)
  - $\gamma = 0$: Solo importa recompensa inmediata (miope)
  - $\gamma \to 1$: Todas las recompensas futuras importan igual (visionario)

---

## 📊 Funciones de Valor

### 1. Función de Valor de Estado: $V^\pi(s)$

**Valor** de estar en estado $s$ siguiendo política $\pi$:

$$V^\pi(s) = \mathbb{E}_\pi[G_t | s_t = s] = \mathbb{E}_\pi \left[ \sum_{k=0}^{\infty} \gamma^k r_{t+k+1} \mid s_t = s \right]$$

**Interpretación**: "¿Qué tan bueno es estar en este estado si sigo mi política $\pi$?"

### 2. Función de Valor de Acción (Q-Function): $Q^\pi(s, a)$

**Valor** de tomar acción $a$ en estado $s$ y luego seguir política $\pi$:

$$Q^\pi(s, a) = \mathbb{E}_\pi[G_t | s_t = s, a_t = a]$$

**Interpretación**: "¿Qué tan bueno es tomar esta acción en este estado?"

### Relación entre $V$ y $Q$

$$V^\pi(s) = \sum_{a \in \mathcal{A}} \pi(a|s) Q^\pi(s, a)$$

---

## 🏆 Política Óptima

### Función de Valor Óptima

$$V^*(s) = \max_\pi V^\pi(s) \quad \text{(mejor valor posible en estado } s \text{)}$$

$$Q^*(s, a) = \max_\pi Q^\pi(s, a) \quad \text{(mejor valor de tomar acción } a \text{ en } s \text{)}$$

### Política Óptima $\pi^*$

Una política es **óptima** si $V^{\pi^*}(s) \geq V^\pi(s)$ para todo $s$ y toda política $\pi$.

Si conocemos $Q^*(s, a)$, la política óptima es:

$$\pi^*(s) = \arg\max_a Q^*(s, a)$$

**Ecuación de Bellman para $Q^*$**:

$$Q^*(s, a) = R(s, a) + \gamma \sum_{s'} P(s'|s, a) \max_{a'} Q^*(s', a')$$

Esta ecuación recursiva es la base de muchos algoritmos de RL.

## 🎲 Exploración vs Explotación

Uno de los dilemas centrales en RL:

* **Explotación** (Exploitation): Elegir la mejor acción conocida hasta ahora
  - Maximiza recompensa inmediata
  - Puede quedar atrapado en óptimos locales

* **Exploración** (Exploration): Probar acciones nuevas
  - Puede descubrir mejores estrategias
  - Sacrifica recompensa a corto plazo

### Estrategias comunes

#### 1. ε-Greedy (Epsilon-Greedy)

Con probabilidad $\epsilon$: elegir acción aleatoria (explorar)  
Con probabilidad $1 - \epsilon$: elegir mejor acción conocida (explotar)

$$
\pi(a|s) = 
\begin{cases}
1 - \epsilon + \frac{\epsilon}{|\mathcal{A}|} & \text{si } a = \arg\max_{a'} Q(s, a') \\
\frac{\epsilon}{|\mathcal{A}|} & \text{en otro caso}
\end{cases}
$$

**Decay de $\epsilon$**: Empezar con $\epsilon$ alto (explorar mucho) y reducirlo gradualmente:

$$\epsilon_t = \epsilon_{\text{min}} + (\epsilon_{\text{max}} - \epsilon_{\text{min}}) e^{-\lambda t}$$

#### 2. Softmax / Boltzmann Exploration

Elegir acciones con probabilidad proporcional a su Q-value:

$$\pi(a|s) = \frac{e^{Q(s,a) / \tau}}{\sum_{a'} e^{Q(s,a') / \tau}}$$

Donde $\tau$ (temperatura) controla cuán "aleatorio" es:
* $\tau \to 0$: Casi determinista (siempre mejor acción)
* $\tau \to \infty$: Uniforme (completamente aleatorio)

#### 3. Upper Confidence Bound (UCB)

Elegir la acción que maximiza:

$$a_t = \arg\max_a \left[ Q(s, a) + c \sqrt{\frac{\ln t}{N(s, a)}} \right]$$

Donde:
* $N(s, a)$: Número de veces que se ha elegido acción $a$ en estado $s$
* $c$: Constante que controla exploración
* El término $\sqrt{\ln t / N(s,a)}$ favorece acciones poco probadas

## 🔧 Métodos de Resolución

Existen tres familias principales de algoritmos:

### 1️⃣ Programación Dinámica (DP)

**Requisito**: Conocer completamente el modelo del MDP ($P$, $R$)

#### Value Iteration

Actualizar iterativamente:

$$V_{k+1}(s) = \max_a \left[ R(s, a) + \gamma \sum_{s'} P(s'|s,a) V_k(s') \right]$$

Hasta convergencia: $V_k \to V^*$

#### Policy Iteration

1. **Policy Evaluation**: Calcular $V^\pi$ para política actual
2. **Policy Improvement**: Mejorar política: $\pi'(s) = \arg\max_a Q^\pi(s, a)$
3. Repetir hasta que $\pi$ no cambie

**Ventaja**: Convergencia garantizada  
**Desventaja**: Requiere modelo completo (raro en práctica)

---

### 2️⃣ Métodos Monte Carlo (MC)

**Idea**: Aprender de **episodios completos** de experiencia

**Algoritmo básico** (First-Visit MC):

1. Generar un episodio siguiendo política $\pi$: $s_0, a_0, r_1, s_1, a_1, r_2, ..., s_T$
2. Para cada estado $s$ visitado:
   - Calcular retorno $G_t = r_{t+1} + \gamma r_{t+2} + ... + \gamma^{T-t-1} r_T$
   - Actualizar: $V(s) \leftarrow V(s) + \alpha [G_t - V(s)]$

Donde $\alpha$ es la tasa de aprendizaje.

**Ventaja**: No necesita modelo del entorno  
**Desventaja**: Solo aprende al final del episodio (puede ser lento)

---

### 3️⃣ Métodos de Diferencia Temporal (TD)

**Idea**: Combinar MC y DP: aprender en **cada paso** usando estimaciones

#### TD(0) - Temporal Difference Learning

Actualización después de cada paso:

$$V(s_t) \leftarrow V(s_t) + \alpha [r_{t+1} + \gamma V(s_{t+1}) - V(s_t)]$$

Donde:
* $r_{t+1} + \gamma V(s_{t+1})$: **TD Target** (estimación del retorno)
* $\delta_t = r_{t+1} + \gamma V(s_{t+1}) - V(s_t)$: **TD Error**

**Ventaja**: Aprende en línea, no necesita episodios completos  
**Desventaja**: Usa estimaciones (puede ser inestable)

## 🧠 Algoritmos Clásicos de RL

### 1. Q-Learning (Off-Policy TD)

**El algoritmo más popular de RL clásico**.

**Actualización**:

$$Q(s_t, a_t) \leftarrow Q(s_t, a_t) + \alpha \left[ r_{t+1} + \gamma \max_{a'} Q(s_{t+1}, a') - Q(s_t, a_t) \right]$$

**Pseudocódigo**:

```
Inicializar Q(s, a) arbitrariamente
Para cada episodio:
    Inicializar s
    Para cada paso del episodio:
        Elegir acción a usando política derivada de Q (ej: ε-greedy)
        Ejecutar a, observar r, s'
        Q(s, a) ← Q(s, a) + α[r + γ max_a' Q(s', a') - Q(s, a)]
        s ← s'
    Fin Para
Fin Para
```

**Características**:
* **Off-policy**: Aprende política óptima aunque use otra para explorar
* **Model-free**: No necesita modelo del entorno
* **Converge a Q^***: Bajo ciertas condiciones (visitar todos (s, a) infinitas veces, $\alpha$ decreciente)

---

### 2. SARSA (On-Policy TD)

**Similar a Q-Learning pero on-policy**.

**Actualización**:

$$Q(s_t, a_t) \leftarrow Q(s_t, a_t) + \alpha \left[ r_{t+1} + \gamma Q(s_{t+1}, a_{t+1}) - Q(s_t, a_t) \right]$$

Donde $a_{t+1}$ es la acción **realmente tomada** (no $\max$).

**Diferencia con Q-Learning**:
* Q-Learning: Usa $\max_{a'} Q(s', a')$ (política óptima)
* SARSA: Usa $Q(s', a')$ donde $a'$ fue realmente elegida (política actual)

**Trade-off**:
* SARSA: Más conservador, aprende la política que realmente sigue
* Q-Learning: Más agresivo, aprende política óptima directamente

---

### 3. Deep Q-Network (DQN)

**Idea**: Aproximar $Q(s, a)$ con una **red neuronal** en vez de tabla.

**Arquitectura**:

```
Estado (s) → [Red Neuronal] → Q(s, a₁), Q(s, a₂), ..., Q(s, aₙ)
```

**Innovaciones clave** (Mnih et al., 2015):

1. **Experience Replay**: Guardar transiciones $(s, a, r, s')$ en buffer, entrenar con minibatches aleatorios
   - Rompe correlaciones entre muestras consecutivas
   - Reutiliza experiencia

2. **Target Network**: Red separada $Q_{\text{target}}$ que se actualiza periódicamente
   - Estabiliza el entrenamiento
   - Evita "perseguir un objetivo móvil"

**Loss function**:

$$L(\theta) = \mathbb{E}_{(s,a,r,s') \sim \mathcal{D}} \left[ \left( r + \gamma \max_{a'} Q_{\text{target}}(s', a'; \theta^-) - Q(s, a; \theta) \right)^2 \right]$$

Donde:
* $\theta$: Parámetros de la red principal
* $\theta^-$: Parámetros de la target network (actualizados cada $C$ pasos)
* $\mathcal{D}$: Replay buffer

**Aplicaciones famosas**: Atari games, robótica, juegos de estrategia

## ⚖️ Ventajas y Desventajas

### ✅ Ventajas

1. **Autonomía**: Aprende sin supervisión directa
2. **Adaptabilidad**: Puede adaptarse a entornos cambiantes
3. **Decisiones secuenciales**: Maneja problemas de planificación a largo plazo
4. **Descubrimiento**: Puede encontrar estrategias no obvias

### ❌ Desventajas

1. **Sample efficiency**: Requiere muchas interacciones con el entorno
2. **Diseño de recompensas**: Difícil definir recompensas correctas
   - **Reward hacking**: Agente explota bugs en recompensas
3. **Estabilidad**: Algoritmos pueden ser inestables (especialmente con redes neuronales)
4. **Exploración**: Puede ser difícil explorar entornos grandes
5. **Costo computacional**: Entrenamiento puede ser muy caro
6. **Simulación vs realidad**: Gap entre simulación y mundo real

## 💡 Conclusión

**Reinforcement Learning** es un paradigma poderoso para problemas de **toma de decisiones secuenciales**. Aunque más complejo que aprendizaje supervisado o no supervisado, permite a los agentes aprender comportamientos sofisticados mediante interacción y experiencia.

### Cuándo usar RL

✅ **Sí**:
* Tienes un entorno donde puedes simular acciones
* Decisiones secuenciales con consecuencias a largo plazo
* Puedes definir recompensas claras
* Exploración es factible

❌ **No**:
* Datos etiquetados abundantes (usa supervisado)
* Una sola decisión (usa clasificación/regresión)
* Recompensas imposibles de definir
* Exploración muy costosa o peligrosa

**¡El aprendizaje por refuerzo está revolucionando la IA, permitiendo agentes autónomos que aprenden por sí mismos!** 🤖✨

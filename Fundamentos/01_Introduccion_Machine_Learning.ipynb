# Introducción al Machine Learning

## 1. ¿Qué es Machine Learning?

**Machine Learning (Aprendizaje Automático)** es una rama de la Inteligencia Artificial que permite a los sistemas aprender y mejorar automáticamente a partir de la experiencia sin ser explícitamente programados.

### Definición Formal

> "Un programa de computadora aprende de la experiencia E con respecto a alguna clase de tareas T y medida de rendimiento P, si su rendimiento en las tareas T, medido por P, mejora con la experiencia E."
>
> — Tom Mitchell (1997)

### Ejemplo

* **Tarea (T)**: Clasificar correos como spam o no spam
* **Experiencia (E)**: Base de datos de correos etiquetados
* **Rendimiento (P)**: Porcentaje de correos clasificados correctamente

### Programación Tradicional vs Machine Learning

| Programación Tradicional | Machine Learning |
|-------------------------|------------------|
| Reglas + Datos → Respuestas | Datos + Respuestas → Reglas |
| Experto define lógica | Algoritmo aprende patrones |
| Difícil para problemas complejos | Eficaz en problemas complejos |

## 2. Tipos de Aprendizaje

### 2.1 Aprendizaje Supervisado

El modelo aprende de **datos etiquetados**: se conocen las entradas y salidas correctas.

$$\mathcal{D} = \{(\mathbf{x}_1, y_1), (\mathbf{x}_2, y_2), ..., (\mathbf{x}_n, y_n)\}$$

**Objetivo:** Aprender función $f: \mathbb{X} \rightarrow \mathbb{Y}$ tal que $f(\mathbf{x}_i) \approx y_i$

**Tipos de problemas:**

* **Clasificación**: $y$ es categórico (spam/no spam, gato/perro/pájaro)
  - Ejemplos: Naive Bayes, Árboles de Decisión, SVM, Random Forest
* **Regresión**: $y$ es continuo (precio, temperatura, edad)
  - Ejemplos: Regresión Lineal, Árboles de Regresión, Gradient Boosting

### 2.2 Aprendizaje No Supervisado

El modelo aprende de **datos sin etiquetas**: solo se conocen las entradas.

$$\mathcal{D} = \{\mathbf{x}_1, \mathbf{x}_2, ..., \mathbf{x}_n\}$$

**Objetivo:** Encontrar estructura oculta en los datos

**Tipos de problemas:**

* **Clustering**: Agrupar datos similares
  - Ejemplos: K-Means, DBSCAN, Hierarchical Clustering
* **Reducción de dimensionalidad**: Comprimir datos manteniendo información
  - Ejemplos: PCA, t-SNE, UMAP
* **Detección de anomalías**: Identificar datos atípicos

### 2.3 Aprendizaje por Refuerzo

Un **agente** aprende a tomar decisiones mediante interacción con un **entorno**.

* Recibe **recompensas** o **penalizaciones** por sus acciones
* Objetivo: Maximizar recompensa acumulada a largo plazo
* Ejemplos: Juegos (AlphaGo), robótica, vehículos autónomos

## 3. Proceso de Machine Learning

### Pipeline de ML

```
1. Definir Problema
        ↓
2. Recolectar Datos
        ↓
3. Explorar y Limpiar Datos (EDA)
        ↓
4. Ingeniería de Características
        ↓
5. Dividir Datos (Train/Validation/Test)
        ↓
6. Seleccionar y Entrenar Modelo
        ↓
7. Evaluar Modelo
        ↓
8. Ajustar Hiperparámetros
        ↓
9. Implementar (Deploy)
        ↓
10. Monitorear y Mantener
```

### División de Datos

* **Training set (60-80%)**: Para entrenar el modelo
* **Validation set (10-20%)**: Para ajustar hiperparámetros
* **Test set (10-20%)**: Para evaluación final

**Importante:** El test set **nunca** se usa durante el entrenamiento.

### Cross-Validation (Validación Cruzada)

Técnica para evaluar modelos dividiendo datos en $k$ particiones:

1. Dividir datos en $k$ folds
2. Para cada fold $i$:
   - Entrenar con $k-1$ folds
   - Validar con fold $i$
3. Promediar resultados

**K-Fold CV** con $k=5$:

```
Fold 1: [Test][Train][Train][Train][Train]
Fold 2: [Train][Test][Train][Train][Train]
Fold 3: [Train][Train][Test][Train][Train]
Fold 4: [Train][Train][Train][Test][Train]
Fold 5: [Train][Train][Train][Train][Test]
```

## 4. Métricas de Evaluación

### 4.1 Clasificación

**Matriz de Confusión:**

|                  | Predicho Positivo | Predicho Negativo |
|------------------|-------------------|-------------------|
| **Real Positivo** | True Positive (TP) | False Negative (FN) |
| **Real Negativo** | False Positive (FP) | True Negative (TN) |

**Métricas:**

$$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$

$$\text{Precision} = \frac{TP}{TP + FP}$$ (De los predichos positivos, ¿cuántos son correctos?)

$$\text{Recall (Sensitivity)} = \frac{TP}{TP + FN}$$ (De los reales positivos, ¿cuántos detectamos?)

$$\text{F1-Score} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$$

**Curva ROC y AUC:**

* ROC (Receiver Operating Characteristic): Grafica TPR vs FPR
* AUC (Area Under Curve): Área bajo la curva ROC
* AUC = 1.0: Clasificador perfecto
* AUC = 0.5: Clasificador aleatorio

### 4.2 Regresión

**Mean Absolute Error (MAE):**

$$MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i|$$

**Mean Squared Error (MSE):**

$$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

**Root Mean Squared Error (RMSE):**

$$RMSE = \sqrt{MSE}$$

**R² (Coeficiente de Determinación):**

$$R^2 = 1 - \frac{SS_{res}}{SS_{tot}}$$

Donde $0 \leq R^2 \leq 1$ (1 = ajuste perfecto)

## 5. Overfitting y Underfitting

### 5.1 Underfitting (Sesgo Alto)

* Modelo **demasiado simple**
* No captura patrones en los datos
* **Alto error** en training y test
* **Solución**: Modelo más complejo, más características

### 5.2 Overfitting (Varianza Alta)

* Modelo **demasiado complejo**
* Memoriza ruido en lugar de patrones generales
* **Bajo error** en training, **alto error** en test
* **Solución**: Regularización, más datos, simplificar modelo

### 5.3 Compromiso Bias-Variance

$$\text{Error Total} = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}$$

* **Bias (Sesgo)**: Error por suposiciones incorrectas del modelo
* **Variance (Varianza)**: Sensibilidad a fluctuaciones en datos de entrenamiento
* **Irreducible Error**: Ruido inherente en los datos

### Trade-off Visual

```
Error ↑
      |
      |        Underfitting      Sweet Spot    Overfitting
      |             \              /     \       /
      |              \            /       \     /
      |               \          /         \   /
      |    Test Error  \________/___________\_/
      |                         /
      |    Train Error _______/___________________
      |______________________________________→ Complejidad
```

### Técnicas de Regularización

* **L1 (Lasso)**: Penaliza suma absoluta de pesos
* **L2 (Ridge)**: Penaliza suma cuadrática de pesos
* **Dropout**: Desactiva neuronas aleatoriamente (redes neuronales)
* **Early Stopping**: Detener entrenamiento antes de overfitting
* **Data Augmentation**: Aumentar datos de entrenamiento

## 6. Ingeniería de Características

Transformar datos crudos en características útiles para el modelo.

### 6.1 Escalamiento

**Normalización (Min-Max Scaling):**

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}} \in [0, 1]$$

**Estandarización (Z-score):**

$$x' = \frac{x - \mu}{\sigma}$$

### 6.2 Codificación de Variables Categóricas

**One-Hot Encoding:**

```
Color: [Rojo, Verde, Azul]
→ Rojo:  [1, 0, 0]
→ Verde: [0, 1, 0]
→ Azul:  [0, 0, 1]
```

**Label Encoding:**

```
Tamaño: [Pequeño, Mediano, Grande]
→ [0, 1, 2]
```

### 6.3 Manejo de Datos Faltantes

* **Eliminar**: Remover filas/columnas con valores faltantes
* **Imputación**: Llenar con media, mediana, moda
* **Modelo predictivo**: Predecir valores faltantes

### 6.4 Creación de Características

* **Polinomiales**: $x^2, x^3, \sqrt{x}$
* **Interacciones**: $x_1 \cdot x_2$
* **Transformaciones**: $\log(x), e^x$
* **Binning**: Discretizar variables continuas

## 7. Conclusiones

4. **Evaluación**: Métricas apropiadas según el problema
5. **Overfitting/Underfitting**: Balance entre complejidad y generalización
6. **Ingeniería de características**: Transformar datos para mejor performance

# Teoría de Regresión

## 1. Introducción

La **Regresión** es una técnica de aprendizaje supervisado para predecir valores continuos.

### Definición Formal

Dado un conjunto de entrenamiento $\\mathcal{D} = \\{(\\mathbf{x}_i, y_i)\\}_{i=1}^{n}$ donde:
- $\\mathbf{x}_i \\in \\mathbb{R}^d$ es el vector de características
- $y_i \\in \\mathbb{R}$ es el valor objetivo continuo

El objetivo es aprender una función:

$$f: \\mathbb{R}^d \\rightarrow \\mathbb{R}$$

Tal que $f(\\mathbf{x}_i) \\approx y_i$ para nuevos datos.

### Ejemplos de Aplicación

* **Precio de viviendas**: $f$(tamaño, ubicación, año) = precio
* **Ventas**: $f$(publicidad, temporada) = ingresos
* **Temperatura**: $f$(hora, mes, ubicación) = temperatura
* **Demanda**: $f$(precio, competencia) = unidades vendidas

## 2. Regresión Lineal Simple

### Modelo

Una variable independiente $x$ predice una variable dependiente $y$:

$$y = \\beta_0 + \\beta_1 x + \\epsilon$$

Donde:
- $\\beta_0$: Intercepto (valor cuando $x=0$)
- $\\beta_1$: Pendiente (cambio en $y$ por unidad de $x$)
- $\\epsilon$: Error aleatorio

### Función de Costo: Error Cuadrático Medio (MSE)

$$J(\\beta_0, \\beta_1) = \\frac{1}{n} \\sum_{i=1}^{n} (y_i - \\hat{y}_i)^2 = \\frac{1}{n} \\sum_{i=1}^{n} (y_i - \\beta_0 - \\beta_1 x_i)^2$$

### Solución: Mínimos Cuadrados

**Derivadas parciales = 0:**

$$\\beta_1 = \\frac{\\sum_{i=1}^{n}(x_i - \\bar{x})(y_i - \\bar{y})}{\\sum_{i=1}^{n}(x_i - \\bar{x})^2} = \\frac{\\text{Cov}(x,y)}{\\text{Var}(x)}$$

$$\\beta_0 = \\bar{y} - \\beta_1 \\bar{x}$$

### Interpretación

* $\\beta_1 > 0$: Relación positiva
* $\\beta_1 < 0$: Relación negativa
* $\\beta_1 = 0$: Sin relación lineal

## 3. Regresión Lineal Múltiple

### Modelo

Múltiples variables independientes:

$$y = \\beta_0 + \\beta_1 x_1 + \\beta_2 x_2 + \\cdots + \\beta_d x_d + \\epsilon$$

**Forma matricial:**

$$\\mathbf{y} = \\mathbf{X}\\boldsymbol{\\beta} + \\boldsymbol{\\epsilon}$$

Donde:
- $\\mathbf{y} = [y_1, y_2, ..., y_n]^T$ (vector $n \\times 1$)
- $\\mathbf{X} = [\\mathbf{1}, \\mathbf{x}_1, \\mathbf{x}_2, ..., \\mathbf{x}_d]$ (matriz $n \\times (d+1)$)
- $\\boldsymbol{\\beta} = [\\beta_0, \\beta_1, ..., \\beta_d]^T$ (parámetros)

### Solución de Mínimos Cuadrados

$$\\boldsymbol{\\hat{\\beta}} = (\\mathbf{X}^T\\mathbf{X})^{-1}\\mathbf{X}^T\\mathbf{y}$$

**Condiciones:**
- $\\mathbf{X}^T\\mathbf{X}$ debe ser invertible
- No multicolinealidad severa
- $n > d$ (más datos que variables)

### Predicción

$$\\hat{y} = \\mathbf{X}\\boldsymbol{\\hat{\\beta}}$$

## 4. Supuestos del Modelo Lineal

### Supuestos Clave

1. **Linealidad**: Relación lineal entre $x$ e $y$
2. **Independencia**: Observaciones independientes
3. **Homoscedasticidad**: Varianza constante de errores
4. **Normalidad**: Errores $\\epsilon \\sim N(0, \\sigma^2)$
5. **No multicolinealidad**: Variables independientes no correlacionadas

### Métricas de Evaluación

**Coeficiente de Determinación ($R^2$):**

$$R^2 = 1 - \\frac{SS_{res}}{SS_{tot}} = 1 - \\frac{\\sum(y_i - \\hat{y}_i)^2}{\\sum(y_i - \\bar{y})^2}$$

* $R^2 = 1$: Ajuste perfecto
* $R^2 = 0$: Modelo no mejor que la media
* $0 \\leq R^2 \\leq 1$

**$R^2$ Ajustado** (penaliza complejidad):

$$R^2_{adj} = 1 - \\frac{(1-R^2)(n-1)}{n-d-1}$$

**RMSE** (Error Cuadrático Medio Raíz):

$$RMSE = \\sqrt{\\frac{1}{n}\\sum_{i=1}^{n}(y_i - \\hat{y}_i)^2}$$

## 5. Regularización

### Problema del Overfitting

Modelos complejos memorizan ruido.

### Ridge Regression (L2)

Añade penalización L2 a la función de costo:

$$J(\\boldsymbol{\\beta}) = \\frac{1}{n}\\sum_{i=1}^{n}(y_i - \\mathbf{x}_i^T\\boldsymbol{\\beta})^2 + \\lambda\\sum_{j=1}^{d}\\beta_j^2$$

$$J(\\boldsymbol{\\beta}) = MSE + \\lambda ||\\boldsymbol{\\beta}||_2^2$$

**Efecto:** Reduce magnitud de coeficientes, no los elimina.

### Lasso Regression (L1)

Penalización L1:

$$J(\\boldsymbol{\\beta}) = \\frac{1}{n}\\sum_{i=1}^{n}(y_i - \\mathbf{x}_i^T\\boldsymbol{\\beta})^2 + \\lambda\\sum_{j=1}^{d}|\\beta_j|$$

**Efecto:** Puede forzar $\\beta_j = 0$ (selección de características).

### Elastic Net

Combina L1 y L2:

$$J(\\boldsymbol{\\beta}) = MSE + \\lambda_1 ||\\boldsymbol{\\beta}||_1 + \\lambda_2 ||\\boldsymbol{\\beta}||_2^2$$

| Método | Penalización | Selección Features | Estabilidad |
|---------|--------------|---------------------|-------------|
| Ridge | L2 | No | Alta |
| Lasso | L1 | Sí | Media |
| Elastic Net | L1 + L2 | Sí | Alta |

## 6. Regresión No Lineal

### Regresión Polinomial

Transforma características a potencias:

$$y = \\beta_0 + \\beta_1 x + \\beta_2 x^2 + \\beta_3 x^3 + \\cdots + \\epsilon$$

**Ejemplo grado 2:**

$$\\mathbf{x} = [x_1, x_2] \\rightarrow [1, x_1, x_2, x_1^2, x_1 x_2, x_2^2]$$

### Árboles de Decisión para Regresión

Particionan espacio y asignan valor promedio por región.

**Criterio de división:** MSE o MAE

### Random Forest Regressor

Ensemble de árboles:

$$\\hat{y} = \\frac{1}{T}\\sum_{t=1}^{T} f_t(\\mathbf{x})$$

### Gradient Boosting Regressor

Combina modelos débiles secuencialmente:

$$f_m(\\mathbf{x}) = f_{m-1}(\\mathbf{x}) + \\alpha h_m(\\mathbf{x})$$

Donde $h_m$ corrige errores de $f_{m-1}$.

## 7. Conclusiones

### Resumen

1. **Regresión Lineal**: Base interpretable
2. **Regularización**: Control de overfitting (Ridge, Lasso)
3. **Métricas**: $R^2$, RMSE, MAE
4. **No lineal**: Polinomial, árboles, ensembles

### Cuándo Usar Cada Método

* **Lineal simple/múltiple**: Relaciones lineales, interpretabilidad
* **Ridge**: Multicolinealidad, mantener todas las variables
* **Lasso**: Selección automática de features
* **Árboles/RF/GB**: No linealidad, interacciones complejas

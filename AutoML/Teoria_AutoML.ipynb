# AutoML y MLOps - Automatización de Machine Learning

## 🎯 Definición

**Automated Machine Learning (AutoML)** es el proceso de **automatizar las tareas** del ciclo de vida de Machine Learning, desde la preparación de datos hasta el despliegue del modelo, permitiendo que usuarios con menos experiencia técnica puedan crear modelos de alta calidad.

**MLOps (Machine Learning Operations)** es el conjunto de prácticas para desplegar, monitorear y mantener modelos de ML en **producción** de forma confiable y escalable.

### La Evolución del ML

```
TRADICIONAL (Manual)              AUTOML (Automatizado)           MLOPS (Producción)
     ↓                                   ↓                              ↓
👨‍💻 Científico de datos          🤖 Automatización               🏭 Operaciones
Codifica cada paso              Sistema decide                  Deploy continuo
Semanas de trabajo              Horas/Minutos                   Monitoreo 24/7
Modelos ad-hoc                  Mejores prácticas               Pipelines robustos
```

---

## 🔄 El Ciclo Completo de ML

### Enfoque Manual (Lo que hemos hecho hasta ahora)

```
1. 📊 Exploración de datos (EDA)
   ├─ Estadísticas descriptivas
   ├─ Visualizaciones
   └─ Detección de outliers
   
2. 🧹 Preprocesamiento
   ├─ Limpieza de datos
   ├─ Imputación de valores faltantes
   ├─ Encoding de categóricas
   └─ Normalización/Estandarización
   
3. 🔧 Feature Engineering
   ├─ Creación de features
   ├─ Selección de features
   └─ Transformaciones
   
4. 🎯 Selección de modelo
   ├─ Probar múltiples algoritmos
   └─ Comparar resultados
   
5. ⚙️ Optimización de hiperparámetros
   ├─ Grid Search
   ├─ Random Search
   └─ Bayesian Optimization
   
6. ✅ Evaluación
   ├─ Métricas
   ├─ Validación cruzada
   └─ Análisis de errores
   
7. 🚀 Despliegue
   └─ Producción

TIEMPO: Días/Semanas
```

### Enfoque AutoML

```
📥 Input: Dataset + Target
    ↓
🤖 AutoML Engine:
    ├─ Preprocesamiento automático
    ├─ Feature engineering automático
    ├─ Prueba de múltiples algoritmos
    ├─ Hyperparameter tuning
    └─ Ensemble de mejores modelos
    ↓
📤 Output: Mejor modelo + Notebook explicativo

TIEMPO: Minutos/Horas
```

## 🤖 ¿Qué Automatiza AutoML?

### 1️⃣ **Preprocesamiento de Datos**

* **Detección automática de tipos**: Numérico, categórico, texto, fecha
* **Imputación inteligente**: 
  - Numéricos: Media, mediana, KNN imputation
  - Categóricos: Moda, nueva categoría "Missing"
* **Encoding automático**:
  - One-Hot Encoding para categóricas de baja cardinalidad
  - Target Encoding para alta cardinalidad
  - Embeddings para texto
* **Scaling automático**: StandardScaler, MinMaxScaler según algoritmo
* **Manejo de outliers**: Detección y tratamiento

### 2️⃣ **Feature Engineering**

* **Creación de features**:
  - Features de fecha/tiempo (día semana, mes, trimestre)
  - Interacciones entre variables
  - Agregaciones estadísticas
* **Selección de features**:
  - Eliminación de correlaciones altas
  - Importancia de features
  - Recursive Feature Elimination (RFE)
* **Transformaciones**:
  - Log, sqrt, polynomial features
  - Binning de variables continuas

### 3️⃣ **Selección de Modelos**

AutoML prueba automáticamente múltiples algoritmos:

**Para Clasificación:**
* Logistic Regression
* Decision Trees
* Random Forest
* Gradient Boosted Trees (XGBoost, LightGBM)
* Neural Networks
* Support Vector Machines

**Para Regresión:**
* Linear Regression (Ridge, Lasso, ElasticNet)
* Decision Tree Regressor
* Random Forest Regressor
* XGBoost, LightGBM Regressor
* Neural Networks

### 4️⃣ **Optimización de Hiperparámetros**

**Métodos de búsqueda:**

#### Grid Search
```python
# Probar todas las combinaciones
params = {
    'n_estimators': [100, 200, 300],
    'max_depth': [5, 10, 15],
    'learning_rate': [0.01, 0.1, 0.3]
}
# Total: 3 × 3 × 3 = 27 combinaciones
```

#### Random Search
```python
# Probar N combinaciones aleatorias
# Más eficiente que Grid Search
# Explora mejor el espacio de hiperparámetros
```

#### Bayesian Optimization ⭐
```python
# El más sofisticado
# Usa resultados previos para decidir qué probar
# Converge más rápido al óptimo
# Usado por Databricks AutoML
```

**Proceso:**

$$\text{Objetivo: Maximizar} \quad f(\theta) = \text{Validation Metric}(\text{Model}_{\theta})$$

Donde $\theta$ son los hiperparámetros.

### 5️⃣ **Ensemble Methods**

AutoML a menudo crea **ensembles** de los mejores modelos:

* **Voting Classifier/Regressor**: Promedio de predicciones
* **Stacking**: Meta-modelo que aprende de otros modelos
* **Weighted Ensemble**: Pesos óptimos para cada modelo

## 🏭 MLOps: De Experimento a Producción

### El Problema del "Research-Production Gap"

```
❌ PROBLEMA COMÚN:

Científico de Datos        →    Ingeniero de Software
"Funciona en mi notebook"       "No puedo desplegarlo"

📓 Jupyter Notebook              🏭 Producción
├─ Código ad-hoc                 ├─ Código robusto
├─ Paths hardcoded               ├─ Configuración
├─ Sin versionado                ├─ Git
├─ Datos locales                 ├─ Pipelines escalables
└─ Un solo modelo                └─ A/B testing
```

### MLOps al Rescate

**MLOps** cierra este gap aplicando prácticas de **DevOps** a ML:

```
🔄 CI/CD para ML:

Code → Train → Test → Deploy → Monitor → Retrain
  ↑                                          ↓
  └──────────────────────────────────────────┘
           (Continuous Improvement)
```

**Qué monitorear:**

  - Relación X → Y cambió
  - Ejemplo: Antes "monto alto" → fraude, ahora no
  
* **Infraestructura**:
  - Latencia de predicciones
  - CPU/Memoria

## ⚖️ AutoML vs Manual: ¿Cuándo usar cada uno?

### ✅ Usar AutoML cuando:

1. **Necesitas un baseline rápido**
   - Proyecto nuevo
   - Exploración inicial
   - Prueba de concepto

2. **Tiempo limitado**
   - Deadline ajustado
   - Múltiples proyectos en paralelo

3. **Equipo con menos experiencia en ML**
   - Analistas de negocio
   - Ingenieros de datos
   - Product managers

4. **Dataset estándar (tabular)**
   - Clasificación binaria/multiclase
   - Regresión
   - No requiere arquitecturas custom

5. **Producción estándar**
   - Predicciones batch o API REST
   - Sin requisitos especiales de latencia

### 🔧 Usar Enfoque Manual cuando:

1. **Problema altamente especializado**
   - Computer Vision custom
   - NLP con arquitectura específica
   - Time Series con patrones únicos
   - Reinforcement Learning

2. **Necesitas control total**
   - Loss function custom
   - Arquitectura de red neuronal específica
   - Ensemble methods complejos

3. **Restricciones estrictas**
   - Latencia ultra-baja (<1ms)
   - Tamaño de modelo muy pequeño (edge devices)
   - Interpretabilidad total requerida

4. **Investigación y publicación**
   - Papers académicos
   - Métodos nuevos
   - Benchmarking riguroso

5. **Cuando AutoML falla**
   - Performance insuficiente
   - Overfitting severo
   - Data drift extremo

### 💡 Mejor Práctica: Enfoque Híbrido

```
1. Ejecutar AutoML primero
   ↓
2. Analizar notebook generado
   ↓  
3. Entender qué funcionó
   ↓
4. Iterar manualmente desde ahí
   ↓
5. Feature engineering custom
   ↓
6. Fine-tuning del mejor modelo
```

**Ejemplo real:**

```
AutoML baseline:  F1 = 0.87
    ↓
Análisis de features importantes
    ↓
Crear features nuevas (manual)
    ↓
Reentrenar con features nuevas
    ↓
Final:  F1 = 0.92  🎉
```

### Flujo End-to-End

```python
# 1. DATA ENGINEERING
raw_data = spark.read.table("bronze.transactions")
cleaned_data = clean_and_transform(raw_data)
cleaned_data.write.saveAsTable("silver.transactions")

# 2. FEATURE ENGINEERING (Feature Store)
from databricks.feature_store import FeatureStoreClient
fs = FeatureStoreClient()

features = compute_features(cleaned_data)
fs.create_table(
    name="features.user_behavior",
    primary_keys=["user_id"],
    df=features
)

# 3. AUTOML TRAINING
from databricks import automl

labels = spark.table("silver.churn_labels")
summary = automl.classify(
    dataset=labels,
    target_col="churned",
    feature_store_lookups=[
        {"table_name": "features.user_behavior", 
         "lookup_key": "user_id"}
    ],
    primary_metric="f1"
)

# 4. MODEL REGISTRY
import mlflow

model_uri = f"runs:/{summary.best_trial.mlflow_run_id}/model"
model_details = mlflow.register_model(model_uri, "churn_model")

# 5. PROMOTE TO PRODUCTION
client = mlflow.tracking.MlflowClient()
client.transition_model_version_stage(
    name="churn_model",
    version=model_details.version,
    stage="Production"
)

# 6. MODEL SERVING (Automatic)
# → Endpoint automáticamente disponible en:
# https://<workspace>.cloud.databricks.com/serving-endpoints/churn_model

# 7. MONITORING
# → Métricas de drift, latencia, throughput automáticas
```

## 💡 Conclusiones y Mejores Prácticas

### La Realidad del ML en la Industria

```
📊 Ciclo de vida del modelo en producción:

10% → Desarrollar modelo inicial
20% → Optimizar y validar
70% → Mantener en producción ⚠️

→ La mayoría del trabajo es MLOps, no desarrollo
```

### Lecciones Clave

#### 1️⃣ **AutoML no reemplaza a los Data Scientists**

✅ **AutoML es excelente para:**
* Baseline rápido
* Exploración inicial
* Producción de modelos "suficientemente buenos"
* Democratización del ML

❌ **AutoML no puede:**
* Definir el problema de negocio
* Decidir qué features son relevantes (requiere domain knowledge)
* Explicar por qué un modelo funciona
* Manejar edge cases complejos
* Innovar en arquitecturas

**AutoML aumenta la productividad, no reemplaza el expertise.**

#### 2️⃣ **MLOps es crucial para el éxito**

```
❌ Sin MLOps:

📓 Jupyter notebook → 💾 Guardado localmente → 🤷 ¿Cómo lo uso?

✅ Con MLOps:

📓 Jupyter → 📦 MLflow → 🏭 Registry → 🚀 Serving → 📊 Monitor
                ↓                                        ↓
         Trazabilidad                            Mejora continua
```

**Estadística clave:**
* 87% de los proyectos de ML nunca llegan a producción
* MLOps reduce este problema dramáticamente

#### 3️⃣ **Feature Engineering sigue siendo el rey**

```
Impacto en performance:

📊 Mejor algoritmo:        +5%  accuracy
⚙️  Mejor hyperparameters:  +3%  accuracy
🔧 Mejores features:       +20% accuracy ⭐
```

**AutoML puede optimizar algoritmos, pero buenos features requieren:**
* Domain knowledge
* Creatividad
* Experimentación

#### 4️⃣ **El modelo es solo una parte del sistema**

```
Sistema ML en producción:

📥 Data ingestion
    ↓
🔄 Data validation
    ↓
🧹 Data preprocessing  
    ↓
🔧 Feature computation
    ↓
🤖 Model inference  ← Solo 5% del código
    ↓
📤 Predictions serving
    ↓
📊 Monitoring & alerting
    ↓
🔄 Retraining pipeline
```

#### 5️⃣ **Monitoreo es tan importante como entrenamiento**

**Modelos se degradan con el tiempo:**

```
Day 1:  Accuracy = 0.95  ✅
Day 30: Accuracy = 0.93  ⚠️
Day 90: Accuracy = 0.87  🚨 Reentrenar
```

**Causas:**
* Data drift (distribución de X cambió)
* Concept drift (relación X→Y cambió)
* Cambios en el negocio
* Estacionalidad

**Solución: Reentrenamiento automático**

```python
if drift_detected() or performance_drop():
    trigger_automl_retraining()
    validate_new_model()
    if new_model_better:
        promote_to_production()
```

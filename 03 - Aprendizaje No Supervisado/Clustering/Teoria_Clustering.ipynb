# Teoría de Clustering

## 1. Introducción

El **Clustering (Agrupamiento)** es una técnica de aprendizaje no supervisado que agrupa datos similares sin etiquetas previas.

### Definición

Dado un conjunto de datos $\mathcal{D} = \{\mathbf{x}_1, \mathbf{x}_2, ..., \mathbf{x}_n\}$ donde $\mathbf{x}_i \in \mathbb{R}^d$:

**Objetivo:** Particionar $\mathcal{D}$ en $k$ grupos (clusters) $C_1, C_2, ..., C_k$ tales que:

* **Cohesión intra-cluster**: Alta similitud dentro de cada grupo
* **Separación inter-cluster**: Baja similitud entre grupos

### Aplicaciones

* **Segmentación de clientes**: Agrupar por comportamiento de compra
* **Compresión de imágenes**: Reducir colores similares
* **Detección de anomalías**: Puntos que no pertenecen a ningún cluster
* **Recomendaciones**: Usuarios con gustos similares
* **Biología**: Taxonomía, genes con funciones similares

## 2. Métricas de Similitud y Distancia

### 2.1 Distancia Euclidiana

$$d(\mathbf{x}, \mathbf{y}) = ||\mathbf{x} - \mathbf{y}||_2 = \sqrt{\sum_{j=1}^{d}(x_j - y_j)^2}$$

**Más común, sensible a escala.**

### 2.2 Distancia de Manhattan

$$d(\mathbf{x}, \mathbf{y}) = ||\mathbf{x} - \mathbf{y}||_1 = \sum_{j=1}^{d}|x_j - y_j|$$

### 2.3 Distancia de Minkowski

Generalización:

$$d(\mathbf{x}, \mathbf{y}) = \left(\sum_{j=1}^{d}|x_j - y_j|^p\right)^{1/p}$$

* $p=1$: Manhattan
* $p=2$: Euclidiana
* $p \rightarrow \infty$: Chebyshev

### 2.4 Similitud Coseno

$$\text{sim}(\mathbf{x}, \mathbf{y}) = \frac{\mathbf{x} \cdot \mathbf{y}}{||\mathbf{x}|| \cdot ||\mathbf{y}||} = \cos(\theta)$$

**Útil para vectores de texto (TF-IDF).**

## 3. K-Means Clustering

### Algoritmo

**Objetivo:** Minimizar la varianza intra-cluster (WCSS):

$$J = \sum_{i=1}^{k} \sum_{\mathbf{x} \in C_i} ||\mathbf{x} - \boldsymbol{\mu}_i||^2$$

Donde $\boldsymbol{\mu}_i$ es el centroide del cluster $i$.

### Algoritmo Lloyd (K-Means estándar)

```python
1. Inicializar: Seleccionar k centroides aleatorios μ1, ..., μk

2. Repetir hasta convergencia:
   a) Asignación: Para cada punto xi:
      C(i) = argmin_j ||xi - μj||^2
   
   b) Actualización: Para cada cluster j:
      μj = (1/|Cj|) ∑_{xi ∈ Cj} xi
   
3. Retornar: Clusters y centroides
```

### Propiedades

* **Convergencia**: Garantizada (puede ser óptimo local)
* **Complejidad**: $O(n \cdot k \cdot d \cdot t)$ donde $t$ = iteraciones
* **Forma de clusters**: Esféricos, similar tamaño
* **Sensibilidad**: A inicialización y outliers

### K-Means++ (Inicialización Inteligente)

1. Elegir primer centroide al azar
2. Para cada siguiente centroide:
   - Calcular $D(x)$ = distancia de cada punto al centroide más cercano
   - Elegir nuevo centroide con probabilidad proporcional a $D(x)^2$
3. Ejecutar K-Means estándar

**Ventaja:** Mejores inicializaciones, menos iteraciones.

## 4. Selección del Número de Clusters (k)

### 4.1 Método del Codo (Elbow Method)

Graficar WCSS vs $k$:

$$WCSS(k) = \sum_{i=1}^{k} \sum_{\mathbf{x} \in C_i} ||\mathbf{x} - \boldsymbol{\mu}_i||^2$$

Buscar "codo" donde la mejora se desacelera.

### 4.2 Coeficiente de Silueta

Para cada punto $\mathbf{x}_i$:

$$s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}$$

Donde:
- $a(i)$: Distancia promedio a puntos en su cluster
- $b(i)$: Distancia promedio al cluster más cercano

**Interpretación:**
- $s(i) \approx 1$: Bien asignado
- $s(i) \approx 0$: En la frontera
- $s(i) < 0$: Mal asignado

**Promedio del conjunto:**

$$\bar{s} = \frac{1}{n}\sum_{i=1}^{n} s(i)$$

Elegir $k$ que maximice $\bar{s}$.

### 4.3 Davies-Bouldin Index

$$DB = \frac{1}{k}\sum_{i=1}^{k} \max_{j \neq i} \frac{\sigma_i + \sigma_j}{d(\boldsymbol{\mu}_i, \boldsymbol{\mu}_j)}$$

**Más bajo es mejor** (clusters compactos y separados).

## 5. Otros Algoritmos de Clustering

### 5.1 Hierarchical Clustering (Jerárquico)

**Aglomerativo (bottom-up):**

1. Empezar: Cada punto es un cluster
2. Repetir: Unir los 2 clusters más cercanos
3. Hasta: Un solo cluster o $k$ clusters

**Linkage (criterio de unión):**

* **Single**: $\min_{\mathbf{x} \in C_i, \mathbf{y} \in C_j} d(\mathbf{x}, \mathbf{y})$
* **Complete**: $\max_{\mathbf{x} \in C_i, \mathbf{y} \in C_j} d(\mathbf{x}, \mathbf{y})$
* **Average**: Promedio de distancias
* **Ward**: Minimiza varianza

**Ventajas:** Dendrograma, no requiere $k$ a priori

**Desventajas:** $O(n^2 \log n)$ o $O(n^3)$

### 5.2 DBSCAN (Density-Based)

Agrupa puntos densos, marca outliers.

**Parámetros:**
- $\epsilon$: Radio de vecindad
- MinPts: Mínimo de puntos para ser denso

**Ventajas:**
* Detecta formas arbitrarias
* Identifica outliers
* No requiere $k$

**Desventajas:**
* Sensible a $\epsilon$ y MinPts
* Problemas con densidades variables

### 5.3 Gaussian Mixture Models (GMM)

Modela datos como mezcla de distribuciones gaussianas:

$$P(\mathbf{x}) = \sum_{i=1}^{k} \pi_i \mathcal{N}(\mathbf{x} | \boldsymbol{\mu}_i, \boldsymbol{\Sigma}_i)$$

**Algoritmo EM (Expectation-Maximization)**

**Ventajas:**
* Clusters elípticos
* Asignación probabilística (soft clustering)
* Fundamento estadístico

**Desventajas:**
* Más costoso que K-Means
* Sensible a inicialización

## 6. Evaluación de Clustering

### Métricas Internas (sin etiquetas)

1. **Coeficiente de Silueta**: $[-1, 1]$, mayor mejor
2. **Davies-Bouldin Index**: $[0, \infty)$, menor mejor
3. **Calinski-Harabasz Index**: Mayor mejor

### Métricas Externas (con etiquetas ground truth)

1. **Adjusted Rand Index (ARI)**: $[-1, 1]$, 1 = perfect match
2. **Normalized Mutual Information (NMI)**: $[0, 1]$, 1 = perfect
3. **V-Measure**: $[0, 1]$, combina homogeneity y completeness

### Consideraciones

* **No hay métrica universal** para elegir el mejor clustering
* Combinar múltiples métricas
* **Validación visual** crucial en 2D/3D
* **Interpretación de dominio** es clave

## 7. Conclusiones

### Resumen

| Algoritmo | Forma Clusters | Requiere k | Outliers | Complejidad |
|-----------|---------------|-----------|----------|-------------|
| K-Means | Esféricos | Sí | No | $O(nkdt)$ |
| Hierarchical | Cualquiera | No | No | $O(n^2 \log n)$ |
| DBSCAN | Arbitraria | No | Sí | $O(n \log n)$ |
| GMM | Elípticos | Sí | No | $O(nkdt)$ |

### Cuándo Usar Cada Método

* **K-Means**: Rápido, clusters esféricos, datos grandes
* **Hierarchical**: Exploración, visualización, datos pequeños
* **DBSCAN**: Formas irregulares, outliers importantes
* **GMM**: Asignación probabilística, modelo generativo

### Consejos Prácticos

1. **Estandarizar datos** antes de clustering
2. **Probar múltiples valores de $k$**
3. **Ejecutar varias veces** (inicializaciones aleatorias)
4. **Visualizar** resultados cuando sea posible
5. **Validar** con conocimiento de dominio
